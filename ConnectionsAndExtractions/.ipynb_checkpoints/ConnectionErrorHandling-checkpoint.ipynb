```python
import mysql.connector
from mysql.connector import Error

try:
    conn = mysql.connector.connect(
        host = "localhost",
        user = "jupyter",
        password = "Wrong",
        database = "sakila"
    )
    if conn.is_connected():
        print("Connected successfully!")
except Error as e:
    print(f"Error: {e}")
finally:
    if 'conn' in locals() and conn.is_connected():
        conn.close()
        print("Connection ended")
```

```python
import time
import mysql.connector
from mysql.connector import Error

MAX_RETRIES= 5
for attempt in range(MAX_RETRIES):
    try:
        conn = mysql.connector.connect(
            host = "localhost",
            user = "jupyter",
            password = "Wrong",
            database = "sakila"
        )
        if conn.is_connected():
          print("Successfully connected on attempt", attempt+1)
          break
    except Error as e:
        print(f"Attempt {attempt +1}: Connection failed due to {e}")
        time.sleep(2 ** attempt) #exponential backoff (2^0, 2^1, 2^2...)
else:
    print("All retry attempts failed. Exiting.")
```

```python
from mysql.connector import errorcode

try:
    conn = mysql.connector.connect(
        host = "localhost",
        user = "jupyter",
        password = "jupPassword",
        database = "wrongDatabase"
    )
    #If connection is successful print message
    if conn.is_connected():
        print("Connected successfully to the 'sakila' database!")

except Error as e:
    if e.errno == errorcode.ER_ACCESS_DENIED_ERROR:
        print("Error: Invalid username or password!")
    elif  e.errno == errorcode.ER_BAD_DB_ERROR:
        print("Error: Database does not exist!")
    else:
        print(f"Unexpected Error: {e}")
        
if 'conn' in locals() and conn.is_connected():
    conn.close()
    print("Database connection closed.")
```
