```python
import mysql.connector

#Establish connection
conn = mysql.connector.connect(
    host="localhost",
    user="jupyter",
    password="jupPassword",
    database="sakila"
)

#check if connection works

if conn.is_connected():
    print("Connected to MySQL database!")

#create cursor object to interact with database
cursor=conn.cursor()
```

```python
#creating a table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees(
        id INT AUTO_INCREMENT PRIMARY KEY,
        name VARCHAR(50),
        department VARCHAR(50),
        salary FLOAT
)
""")
conn.commit()
print("Table created successfully!")
```

```python
#insterting into a table
sql_insert= "INSERT INTO employees (name, department, salary) VALUES (%s, %s, %s)"
values = [
    ("Allyson", "Copywriting", 80000),
    ("Cypress", "IT", 70000),
    ("Honey", "Morale", 100000)
]

#insert multiple rows
cursor.executemany(sql_insert, values) 
#commit the transaction
conn.commit()
print("Data inserted successfully")
```

```python
#fetch rows
cursor.execute("SELECT * FROM employees")
rows = cursor.fetchall()

#display rows
for row in rows:
    print(row)
```

```python
#Updating
sql_update = "UPDATE employees SET salary = salary + 5000 WHERE department = 'IT'"
cursor.execute(sql_update)
conn.commit()
print("Salary updated for IT department")
```

```python
cursor.execute("SELECT * FROM employees WHERE department = 'IT'")
print(cursor.fetchall())
```

```python
#Delete rows
sql_delete = "DELETE FROM employees WHERE id > 3"
cursor.execute(sql_delete)
conn.commit()
print("Record successfully deleted")
```

```python
cursor.execute("SELECT * FROM employees")
print(cursor.fetchall())
```

```python
try:
    cursor.execute("UPDATE employees SET salary = salary - 5000 WHERE department = 'IT'")
    raise Exception("Simulated Error!") #simulating an issue
    conn.commit() #shouldn't commit because of the error
except Exception as e:
    conn.rollback() #rolls back previous state
    print("Transaction rolled back due to error:", e)
```

```python
cursor.close()
conn.close()
print("Database connection closed!")
```
